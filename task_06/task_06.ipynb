```python
import re
```

```python
MONTHS_RU = [
    "янв", "январь", "января",
    "фев", "февраль", "февраля",
    "мар", "март", "марта",
    "апр", "апрель", "апреля",
    "май", "мая",
    "июн", "июнь", "июня",
    "июл", "июль", "июля",
    "авг", "август", "августа",
    "сен", "сентябрь", "сентября",
    "окт", "октябрь", "октября",
    "ноя", "ноябрь", "ноября",
    "дек", "декабрь", "декабря"
]


MONTHS_EN = [
    "jan", "january", "feb", "february", "mar", "march",
    "apr", "april", "may", "jun", "june", "jul", "july",
    "aug", "august", "sep", "september", "oct", "october",
    "nov", "november", "dec", "december"
]

ALL_MONTHS = MONTHS_RU + MONTHS_EN
```

```python
def make_month_group(months_list):
    escaped = [re.escape(m) for m in months_list]
    return "(" + "|".join(escaped) + ")"
```

```python
months_pattern = make_month_group(ALL_MONTHS)
```

```python
patterns = [
    r"\b\d{1,2}[\s.\/-]\d{1,2}[\s.\/-]\d{2,4}\b",
    r"\b\d{1,2}[\s.\/-]\d{1,2}[\s.\/-]\d{2,4}\b",
    r"\b\d{4}[\s.\/-]\d{1,2}[\s.\/-]\d{1,2}\b",
    rf"\b\d{{4}}[\s.\/-]{months_pattern}[\s.\/-]\d{{1,2}}\b",      # YYYY Month DD
    rf"\b\d{{1,2}}[\s.\/-]{months_pattern}[\s.\/-]\d{{2,4}}\b",    # DD Month YYYY
    rf"\b{months_pattern}[\s.\/-]\d{{1,2}},?[\s.\/-]*\d{{2,4}}\b",  # Month DD, YYYY
]
```

```python
test_dates = [
    "01.01.2024", "31/12/2023", "5-10-2022", "2024-03-15", "23.02.23",
    "5 января 2024", "01 янв 23", "31 декабря 1999", "фев 5 2005",
    "January 5, 2024", "Jan 5 2024", "December 31, 1999",
    "2024 январь 5", "2023 Jan 1", "январь 5, 2024", "Jan 5, 2024",
    "01-янв-2024", "Jan/5/2024", "2024.фев.10", "5.мая.2020",
    "28.02.2023", "29.02.2024", "1.1.00", "2024-1-1",
    "ЯНВАРЬ 5 2024", "Январь 5 2024", "JAN 5 2024",
    # invalid
    "32.01.2024", "30.02.2023", "40 декабря 2020", "Feb 30 2021"
]
```

```python
for date in test_dates:
    matched = False
    for p in patterns:
        if re.search(p, date, flags=re.IGNORECASE):
            matched = True
            break
    status = "MATCHED" if matched else "NO MATCH"
    print(f"{status} | {date}")
```
